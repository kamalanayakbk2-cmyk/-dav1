```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# IQR calculation
Q1 = df['math score'].quantile(0.25)
Q3 = df['math score'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# IQR capping
df['Math_IQR_Capped'] = df['math score'].clip(
    lower=lower_limit,
    upper=upper_limit
)

# Z-score capping
mean = df['math score'].mean()
std = df['math score'].std()

lower_z_limit = mean - 3 * std
upper_z_limit = mean + 3 * std

df['Math_Z_Capped'] = df['math score'].clip(
    lower=lower_z_limit,
    upper=upper_z_limit
)

# Comparison graph
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[
    ['math score', 'Math_IQR_Capped', 'Math_Z_Capped']
])

plt.title("Comparison Before and After Outlier Treatment")
plt.ylabel("Math Score")
plt.tight_layout()
plt.show()
```
