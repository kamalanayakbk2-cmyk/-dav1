```python
import numpy as np
import pandas as pd
```

```python
a= np.array([13, 34, 76, 23,48, 55])
print('Array: ',a)
print('Mean:', a.mean())
print('Reshaped 2D array: ', a.reshape(3,2))
print('Sliced: ', a[1:5])
```

```python
data= { 'Name' : ['Ram', 'Sham', 'Ravi'],
        'Marks' : [60, 76, 59],
        'Branch' : ['CS', 'ME', 'EC']}
df= pd.DataFrame(data)
        
```

```python
print(df)
print("\n First 2 rows: \n", df.iloc[0:2])
print("\n Marks column: \n", df['Marks'])
print("\n Rows where Marks>70 :\n", df[df['Marks']>70])
```
