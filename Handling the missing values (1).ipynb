```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\jjksa\Downloads\Titanic-Dataset.csv")
```

```python
print("Original Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
print("Missing Values:")
print(df.isnull())

print("\nNumber of Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Mean imputation")
```

```python
df_mean = df.copy()

df_mean['Age'] = df_mean['Age'].fillna(df_mean['Age'].mean())

print("Dataset after Mean Imputation:")
print(df_mean.head(10))

print("\nMean Age:")
print(df_mean['Age'].mean())
```

```python
print("Median imputation")
```

```python
df_median = df.copy()

df_median['Fare'] = df_median['Fare'].fillna(df_median['Fare'].median())

print("Dataset after Median Imputation:")
print(df_median.head(10))

print("\nMedian Fare:")
print(df_median['Fare'].median())
```

```python
print("Forward fill")
```

```python
df_ffill = df.copy()

print("Before Forward Fill:")
print(df_ffill[['Embarked']].head(10))

df_ffill['Embarked'] = df_ffill['Embarked'].ffill()

print("\nAfter Forward Fill:")
print(df_ffill[['Embarked']].head(10))
```

```python
print(": Label encoding")
```

```python
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()

df_label['Sex'] = df_label['Sex'].fillna(df_label['Sex'].mode()[0])

encoder = LabelEncoder()

df_label['Sex_Encoded'] = encoder.fit_transform(df_label['Sex'])

print("Label Encoding:")
print(df_label[['Sex', 'Sex_Encoded']].head(10))

print("\nClasses:")
print(encoder.classes_)
```

```python
print("One-hot encoding")
```

```python
df_onehot = df.copy()

df_onehot['Embarked'] = df_onehot['Embarked'].fillna(
    df_onehot['Embarked'].mode()[0]
)

encoded_df = pd.get_dummies(
    df_onehot,
    columns=['Embarked'],
    dtype=int
)

print("One-Hot Encoded Dataset:")
print(encoded_df.head())
```
