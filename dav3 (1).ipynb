```python
from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df.boxplot()

plt.title("Box Plot of Iris Dataset")
plt.xlabel("Features")
plt.ylabel("Values")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Draw box plot
df.boxplot(column='sepal length (cm)')

# Display graph
plt.title("Box Plot of Iris Dataset")
plt.ylabel("Sepal Length (cm)")
plt.show()
```
