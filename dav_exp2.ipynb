```python
# Load the Iris dataset
data = pd.read_csv("Iris (1).csv")
```

```python
print(data.columns)
```

```python
# Extract petal length values
petal_length = data["PetalLengthCm"].to_numpy()
```

```python
# Calculate statistical values
highest = np.max(petal_length)
lowest = np.min(petal_length)
average = np.mean(petal_length)
```

```python

# Display results
print("Highest Petal Length:", highest)
print("Lowest Petal Length:", lowest)
print("Average Petal Length:", average)
```
